```python
# ================================================================
# SMART CHASSIS VIBRATION ANALYSIS SYSTEM
# MACHINE LEARNING PROJECT
# Dataset: PVS - Passive Vehicular Sensors
# ================================================================

# =========================
# 1. INSTALL LIBRARIES
# =========================

!pip -q install kagglehub xgboost scipy joblib

# =========================
# 2. IMPORT LIBRARIES
# =========================

import os
import glob
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.fft import rfft, rfftfreq
from scipy.stats import skew, kurtosis

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

print("=" * 70)
print("SMART CHASSIS VIBRATION ANALYSIS SYSTEM")
print("Machine Learning Pipeline")
print("=" * 70)


# =========================
# 3. DOWNLOAD PVS DATASET
# =========================

import kagglehub

print("\nDownloading PVS dataset...")

dataset_path = kagglehub.dataset_download(
    "jefmenegazzo/pvs-passive-vehicular-sensors-datasets"
)

print("Dataset downloaded to:")
print(dataset_path)


# =========================
# 4. FIND ALL CSV FILES
# =========================

csv_files = glob.glob(
    os.path.join(dataset_path, "**", "*.csv"),
    recursive=True
)

print("\nNumber of CSV files found:", len(csv_files))

for file in csv_files:
    print(os.path.basename(file))


# =========================
# 5. FIND MPU AND LABEL FILES
# =========================

mpu_files = [
    f for f in csv_files
    if "mpu" in os.path.basename(f).lower()
]

label_files = [
    f for f in csv_files
    if "label" in os.path.basename(f).lower()
]

print("\nMPU files:")
for f in mpu_files:
    print(f)

print("\nLabel files:")
for f in label_files:
    print(f)


# =========================
# 6. READ FIRST MPU FILE
# =========================

if len(mpu_files) == 0:
    raise FileNotFoundError(
        "No MPU accelerometer CSV files were found."
    )

mpu_file = mpu_files[0]

df = pd.read_csv(mpu_file)

print("\nUsing MPU file:")
print(mpu_file)

print("\nShape:", df.shape)

print("\nColumns:")
for i, col in enumerate(df.columns):
    print(i, ":", col)

print("\nFirst five rows:")
print(df.head())


# =========================
# 7. AUTOMATICALLY FIND
#    ACCELERATION COLUMNS
# =========================

columns_lower = {
    str(c).lower(): c for c in df.columns
}

acc_columns = []

for col in df.columns:

    name = str(col).lower()

    if (
        "acc" in name
        or "acceler" in name
    ):
        acc_columns.append(col)

print("\nPossible acceleration columns:")
print(acc_columns)


# ------------------------------------------------
# If automatic detection does not find 3 columns,
# print all columns and stop so they can be selected.
# ------------------------------------------------

if len(acc_columns) < 3:

    print("\nWARNING:")
    print("Could not automatically identify 3 acceleration columns.")
    print("Available columns are:")
    print(df.columns.tolist())

    raise ValueError(
        "Please inspect the printed column names and select "
        "the X, Y and Z acceleration columns."
    )


# Select first three acceleration columns
acc_columns = acc_columns[:3]

print("\nSelected acceleration columns:")
print(acc_columns)


# =========================
# 8. CLEAN ACCELERATION DATA
# =========================

data = df[acc_columns].copy()

data.columns = [
    "acc_x",
    "acc_y",
    "acc_z"
]

data = data.apply(
    pd.to_numeric,
    errors="coerce"
)

data = data.replace(
    [np.inf, -np.inf],
    np.nan
)

data = data.dropna()

print("\nCleaned acceleration data:")
print(data.shape)


# =========================
# 9. CALCULATE VIBRATION
#    MAGNITUDE
# =========================

data["acc_magnitude"] = np.sqrt(
    data["acc_x"] ** 2 +
    data["acc_y"] ** 2 +
    data["acc_z"] ** 2
)

print("\nAcceleration magnitude calculated.")

print(
    data["acc_magnitude"].describe()
)


# =========================
# 10. RAW VIBRATION GRAPH
# =========================

plt.figure(figsize=(14, 5))

N_plot = min(
    5000,
    len(data)
)

plt.plot(
    data["acc_magnitude"].iloc[:N_plot]
)

plt.title(
    "Vehicle Chassis/Body Vibration Signal"
)

plt.xlabel("Sample")
plt.ylabel("Acceleration Magnitude")

plt.grid(True)
plt.tight_layout()
plt.show()


# =========================
# 11. FEATURE EXTRACTION
# =========================

FS = 100          # PVS sampling frequency
WINDOW_SIZE = 200 # 2 seconds at 100 Hz

def extract_features(signal, fs=100):

    signal = np.asarray(signal)

    features = {}

    # -------------------------
    # TIME DOMAIN
    # -------------------------

    features["mean"] = np.mean(signal)

    features["std"] = np.std(signal)

    features["rms"] = np.sqrt(
        np.mean(signal ** 2)
    )

    features["peak"] = np.max(
        np.abs(signal)
    )

    features["peak_to_peak"] = (
        np.max(signal) -
        np.min(signal)
    )

    features["skewness"] = skew(
        signal
    )

    features["kurtosis"] = kurtosis(
        signal
    )

    # Crest factor
    if features["rms"] != 0:

        features["crest_factor"] = (
            features["peak"] /
            features["rms"]
        )

    else:

        features["crest_factor"] = 0

    # -------------------------
    # FREQUENCY DOMAIN
    # -------------------------

    fft_values = np.abs(
        rfft(signal)
    )

    frequencies = rfftfreq(
        len(signal),
        1 / fs
    )

    if len(fft_values) > 1:

        # Ignore DC component
        dominant_idx = (
            np.argmax(
                fft_values[1:]
            ) + 1
        )

        features["dominant_frequency"] = (
            frequencies[dominant_idx]
        )

        features["spectral_energy"] = (
            np.sum(
                fft_values ** 2
            )
        )

        features["max_fft_amplitude"] = (
            np.max(
                fft_values[1:]
            )
        )

    else:

        features["dominant_frequency"] = 0

        features["spectral_energy"] = 0

        features["max_fft_amplitude"] = 0

    return features


# =========================
# 12. CREATE FEATURE DATASET
# =========================

feature_rows = []

signal = data[
    "acc_magnitude"
].values

for start in range(
    0,
    len(signal) - WINDOW_SIZE,
    WINDOW_SIZE
):

    window = signal[
        start:start + WINDOW_SIZE
    ]

    features = extract_features(
        window,
        FS
    )

    feature_rows.append(
        features
    )

feature_df = pd.DataFrame(
    feature_rows
)

print("\nFeature dataset:")
print(feature_df.shape)

print(
    feature_df.head()
)


# =========================
# 13. FIND LABEL DATA
# =========================

if len(label_files) == 0:

    raise FileNotFoundError(
        "No label CSV file found."
    )

label_file = label_files[0]

labels_df = pd.read_csv(
    label_file
)

print("\nLabel file:")
print(label_file)

print("\nLabel columns:")
print(
    labels_df.columns.tolist()
)

print("\nFirst rows:")
print(
    labels_df.head()
)


# ============================================================
# 14. AUTOMATIC LABEL SEARCH
# ============================================================

label_column = None

preferred_words = [
    "label",
    "class",
    "condition",
    "road",
    "roughness"
]

for word in preferred_words:

    for col in labels_df.columns:

        if word in str(col).lower():

            # Make sure it is not obviously a numeric time column
            if labels_df[col].dtype == "object":

                label_column = col
                break

    if label_column is not None:
        break


# If no categorical label was detected,
# inspect unique values.

if label_column is None:

    print(
        "\nNo obvious categorical label column "
        "was automatically identified."
    )

    print(
        "Unique values for each column:"
    )

    for col in labels_df.columns:

        try:

            print(
                col,
                "->",
                labels_df[col]
                .dropna()
                .unique()[:20]
            )

        except:
            pass

    raise ValueError(
        "Please identify the correct label column."
    )


print(
    "\nSelected label column:",
    label_column
)


# =========================
# 15. CREATE WINDOW LABELS
# =========================

raw_labels = labels_df[
    label_column
].dropna().astype(str).values


# ------------------------------------------------------------
# We need one label per vibration window.
# The safest generic approach is proportional alignment.
# ------------------------------------------------------------

n_windows = len(feature_df)

if len(raw_labels) >= n_windows:

    # If labels already correspond to windows
    window_labels = raw_labels[
        :n_windows
    ]

else:

    # If labels are shorter, interpolate their indices
    indices = np.linspace(
        0,
        len(raw_labels) - 1,
        n_windows
    ).astype(int)

    window_labels = raw_labels[
        indices
    ]


feature_df["label"] = window_labels


# =========================
# 16. CLEAN LABELS
# =========================

feature_df = feature_df.dropna(
    subset=["label"]
)

feature_df["label"] = (
    feature_df["label"]
    .astype(str)
    .str.strip()
)


print(
    "\nClass distribution:"
)

print(
    feature_df["label"]
    .value_counts()
)


# =========================
# 17. REMOVE CLASSES WITH
#     TOO FEW SAMPLES
# =========================

class_counts = (
    feature_df["label"]
    .value_counts()
)

valid_classes = class_counts[
    class_counts >= 5
].index

feature_df = feature_df[
    feature_df["label"]
    .isin(valid_classes)
].reset_index(drop=True)


print(
    "\nClasses used for ML:"
)

print(
    feature_df["label"]
    .value_counts()
)


# =========================
# 18. PREPARE X AND Y
# =========================

X = feature_df.drop(
    columns=["label"]
)

y_text = feature_df[
    "label"
]


# Encode labels
label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    y_text
)

print(
    "\nEncoded classes:"
)

for number, name in enumerate(
    label_encoder.classes_
):

    print(
        number,
        "->",
        name
    )


# =========================
# 19. TRAIN TEST SPLIT
# =========================

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
)


print(
    "\nTraining samples:",
    len(X_train)
)

print(
    "Testing samples:",
    len(X_test)
)


# =========================
# 20. SCALE DATA FOR SVM
# =========================

scaler = StandardScaler()

X_train_scaled = (
    scaler.fit_transform(
        X_train
    )
)

X_test_scaled = (
    scaler.transform(
        X_test
    )
)


# ================================================================
# 21. RANDOM FOREST
# ================================================================

print("\n" + "=" * 70)
print("RANDOM FOREST")
print("=" * 70)

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf.fit(
    X_train,
    y_train
)

rf_pred = rf.predict(
    X_test
)

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_precision = precision_score(
    y_test,
    rf_pred,
    average="weighted",
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_pred,
    average="weighted",
    zero_division=0
)

rf_f1 = f1_score(
    y_test,
    rf_pred,
    average="weighted",
    zero_division=0
)

print(
    classification_report(
        y_test,
        rf_pred,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ================================================================
# 22. SVM
# ================================================================

print("\n" + "=" * 70)
print("SUPPORT VECTOR MACHINE")
print("=" * 70)

svm = SVC(
    kernel="rbf",
    C=10,
    gamma="scale"
)

svm.fit(
    X_train_scaled,
    y_train
)

svm_pred = svm.predict(
    X_test_scaled
)

svm_accuracy = accuracy_score(
    y_test,
    svm_pred
)

svm_precision = precision_score(
    y_test,
    svm_pred,
    average="weighted",
    zero_division=0
)

svm_recall = recall_score(
    y_test,
    svm_pred,
    average="weighted",
    zero_division=0
)

svm_f1 = f1_score(
    y_test,
    svm_pred,
    average="weighted",
    zero_division=0
)

print(
    classification_report(
        y_test,
        svm_pred,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ================================================================
# 23. XGBOOST
# ================================================================

print("\n" + "=" * 70)
print("XGBOOST")
print("=" * 70)

num_classes = len(
    np.unique(y)
)

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="mlogloss",
    objective="multi:softprob",
    num_class=num_classes
)

xgb.fit(
    X_train,
    y_train
)

xgb_pred = xgb.predict(
    X_test
)

xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)

xgb_precision = precision_score(
    y_test,
    xgb_pred,
    average="weighted",
    zero_division=0
)

xgb_recall = recall_score(
    y_test,
    xgb_pred,
    average="weighted",
    zero_division=0
)

xgb_f1 = f1_score(
    y_test,
    xgb_pred,
    average="weighted",
    zero_division=0
)

print(
    classification_report(
        y_test,
        xgb_pred,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ================================================================
# 24. MODEL COMPARISON
# ================================================================

results = pd.DataFrame({

    "Model": [
        "Random Forest",
        "SVM",
        "XGBoost"
    ],

    "Accuracy": [
        rf_accuracy,
        svm_accuracy,
        xgb_accuracy
    ],

    "Precision": [
        rf_precision,
        svm_precision,
        xgb_precision
    ],

    "Recall": [
        rf_recall,
        svm_recall,
        xgb_recall
    ],

    "F1 Score": [
        rf_f1,
        svm_f1,
        xgb_f1
    ]
})


print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

print(
    results.to_string(
        index=False
    )
)


# =========================
# 25. PERFORMANCE GRAPH
# =========================

results_plot = results.set_index(
    "Model"
)

results_plot.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title(
    "Machine Learning Model Performance"
)

plt.ylabel(
    "Score"
)

plt.xlabel(
    "Machine Learning Model"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y"
)

plt.tight_layout()
plt.show()


# ================================================================
# 26. SELECT BEST MODEL
# ================================================================

best_index = results[
    "F1 Score"
].idxmax()

best_model_name = results.loc[
    best_index,
    "Model"
]

print(
    "\nBest model based on F1 Score:",
    best_model_name
)


if best_model_name == "Random Forest":

    best_model = rf
    best_prediction = rf_pred

elif best_model_name == "SVM":

    best_model = svm
    best_prediction = svm_pred

else:

    best_model = xgb
    best_prediction = xgb_pred


# ================================================================
# 27. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    best_prediction
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=label_encoder.classes_
)

fig, ax = plt.subplots(
    figsize=(8,7)
)

disp.plot(
    ax=ax,
    cmap="Blues"
)

plt.title(
    "Confusion Matrix - " +
    best_model_name
)

plt.tight_layout()
plt.show()


# ================================================================
# 28. FEATURE IMPORTANCE
# ================================================================

if best_model_name in [
    "Random Forest",
    "XGBoost"
]:

    importance = pd.Series(
        best_model.feature_importances_,
        index=X.columns
    ).sort_values(
        ascending=False
    )

    plt.figure(
        figsize=(10,6)
    )

    importance.plot(
        kind="bar"
    )

    plt.title(
        "Feature Importance - " +
        best_model_name
    )

    plt.xlabel(
        "Vibration Feature"
    )

    plt.ylabel(
        "Importance"
    )

    plt.xticks(
        rotation=45,
        ha="right"
    )

    plt.grid(
        axis="y"
    )

    plt.tight_layout()
    plt.show()

    print(
        "\nFeature importance:"
    )

    print(
        importance
    )


# ================================================================
# 29. FFT EXAMPLE
# ================================================================

example_signal = signal[
    :WINDOW_SIZE
]

fft_values = np.abs(
    rfft(example_signal)
)

frequencies = rfftfreq(
    WINDOW_SIZE,
    1 / FS
)

plt.figure(
    figsize=(12,5)
)

plt.plot(
    frequencies,
    fft_values
)

plt.title(
    "Frequency Spectrum of Vehicle Vibration"
)

plt.xlabel(
    "Frequency (Hz)"
)

plt.ylabel(
    "FFT Amplitude"
)

plt.xlim(
    0,
    FS / 2
)

plt.grid(True)

plt.tight_layout()
plt.show()


# ================================================================
# 30. SAVE MODEL
# ================================================================

model_package = {

    "model": best_model,

    "scaler": scaler,

    "label_encoder": label_encoder,

    "features": X.columns.tolist(),

    "sampling_frequency": FS,

    "window_size": WINDOW_SIZE

}

joblib.dump(
    model_package,
    "smart_chassis_vibration_model.pkl"
)

print(
    "\nSaved model:"
)

print(
    "smart_chassis_vibration_model.pkl"
)


# ================================================================
# 31. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)

print(
    "Dataset: PVS Passive Vehicular Sensors"
)

print(
    "Sampling frequency:",
    FS,
    "Hz"
)

print(
    "Window size:",
    WINDOW_SIZE,
    "samples"
)

print(
    "Number of features:",
    X.shape[1]
)

print(
    "Number of classes:",
    len(label_encoder.classes_)
)

print(
    "Classes:",
    list(label_encoder.classes_)
)

print(
    "\nRandom Forest Accuracy:",
    round(rf_accuracy * 100, 2),
    "%"
)

print(
    "SVM Accuracy:",
    round(svm_accuracy * 100, 2),
    "%"
)

print(
    "XGBoost Accuracy:",
    round(xgb_accuracy * 100, 2),
    "%"
)

print(
    "\nBEST MODEL:",
    best_model_name
)

print("=" * 70)
print("PROJECT COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# 20. PVS LABEL PROCESSING
# ================================================================

print("\n" + "=" * 75)
print("PVS ROAD-CONDITION LABEL PROCESSING")
print("=" * 75)

# ------------------------------------------------
# PVS LEFT-SIDE LABELS
# ------------------------------------------------

LEFT_LABEL_COLUMNS = [
    "good_road_left",
    "regular_road_left",
    "bad_road_left"
]

# Check that the required columns exist

missing_columns = [
    col
    for col in LEFT_LABEL_COLUMNS
    if col not in labels.columns
]

if len(missing_columns) > 0:

    raise ValueError(
        f"Missing label columns: {missing_columns}"
    )

print("\nUsing LEFT-side labels:")
print(LEFT_LABEL_COLUMNS)


# ================================================================
# 21. INSPECT LABEL DATA
# ================================================================

print("\nLabel dataset shape:")
print(labels.shape)

print("\nLabel sample:")

display(
    labels[
        LEFT_LABEL_COLUMNS
    ].head(10)
)


print("\nLabel totals:")

print(
    labels[
        LEFT_LABEL_COLUMNS
    ].sum()
)


# ================================================================
# 22. CONVERT ONE-HOT LABELS INTO CLASS LABEL
# ================================================================

def convert_one_hot_to_class(row):

    values = np.asarray(
        row.values,
        dtype=float
    )

    # If no label is active
    if np.sum(values) == 0:

        return "unknown"

    # Select the active class
    index = np.argmax(
        values
    )

    class_names = [
        "good_road",
        "regular_road",
        "bad_road"
    ]

    return class_names[index]


raw_class_labels = (
    labels[
        LEFT_LABEL_COLUMNS
    ]
    .apply(
        convert_one_hot_to_class,
        axis=1
    )
)


print("\nConverted classes:")

print(
    raw_class_labels.value_counts()
)


# ================================================================
# 23. ALIGN LABELS WITH 2-SECOND WINDOWS
# ================================================================

# Our vibration feature window:
#
# Sampling rate = 100 Hz
# Window = 2 seconds
# Samples/window = 200


number_of_feature_windows = len(
    feature_df
)


print(
    "\nNumber of vibration windows:",
    number_of_feature_windows
)


print(
    "Number of raw label samples:",
    len(raw_class_labels)
)


# ------------------------------------------------
# CASE 1:
# Label data corresponds to raw sensor samples
# ------------------------------------------------

if len(raw_class_labels) >= (
    number_of_feature_windows *
    WINDOW_SIZE
):

    print(
        "\nAligning labels using the same "
        "2-second windows as vibration data."
    )

    window_labels = []

    for start in range(
        0,
        number_of_feature_windows *
        WINDOW_SIZE,
        WINDOW_SIZE
    ):

        end = (
            start +
            WINDOW_SIZE
        )

        current_labels = (
            raw_class_labels.iloc[
                start:end
            ]
        )

        # Remove unknown labels
        current_labels = (
            current_labels[
                current_labels != "unknown"
            ]
        )

        if len(current_labels) == 0:

            window_labels.append(
                "unknown"
            )

        else:

            # Majority class within window
            window_labels.append(
                current_labels
                .value_counts()
                .idxmax()
            )


# ------------------------------------------------
# CASE 2:
# Labels are already approximately window-level
# ------------------------------------------------

else:

    print(
        "\nLabel file is shorter than the "
        "raw-sample requirement."
    )

    print(
        "Using proportional alignment."
    )

    indices = np.linspace(
        0,
        len(raw_class_labels) - 1,
        number_of_feature_windows
    ).astype(int)

    window_labels = (
        raw_class_labels.iloc[
            indices
        ].values
    )


# ================================================================
# 24. ADD LABELS TO FEATURE DATASET
# ================================================================

feature_df[
    "label"
] = window_labels


# ================================================================
# 25. REMOVE UNKNOWN LABELS
# ================================================================

feature_df = feature_df[
    feature_df[
        "label"
    ] != "unknown"
].reset_index(
    drop=True
)


print(
    "\nFinal labelled feature dataset:"
)

print(
    feature_df.shape
)


# ================================================================
# 26. CLASS DISTRIBUTION
# ================================================================

print(
    "\nFinal class distribution:"
)

class_distribution = (
    feature_df[
        "label"
    ]
    .value_counts()
)

print(
    class_distribution
)


# ================================================================
# 27. CLASS DISTRIBUTION GRAPH
# ================================================================

plt.figure(
    figsize=(8, 5)
)

class_distribution.plot(
    kind="bar"
)

plt.title(
    "Vehicle Vibration Class Distribution"
)

plt.xlabel(
    "Road Condition"
)

plt.ylabel(
    "Number of 2-Second Windows"
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ================================================================
# 28. PREPARE X AND Y
# ================================================================

X = feature_df.drop(
    columns=[
        "label"
    ]
)

y_text = feature_df[
    "label"
]


# ================================================================
# 29. ENCODE CLASSES
# ================================================================

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    y_text
)


print(
    "\nEncoded classes:"
)

for i, class_name in enumerate(
    label_encoder.classes_
):

    print(
        i,
        "->",
        class_name
    )


# ================================================================
# 30. TRAIN / TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
)


print(
    "\nTraining samples:",
    len(X_train)
)

print(
    "Testing samples:",
    len(X_test)
)


# ================================================================
# 31. STANDARDIZATION
# ================================================================

scaler = StandardScaler()

X_train_scaled = (
    scaler.fit_transform(
        X_train
    )
)

X_test_scaled = (
    scaler.transform(
        X_test
    )
)


# ================================================================
# 32. RANDOM FOREST
# ================================================================

print("\n" + "=" * 75)
print("RANDOM FOREST")
print("=" * 75)

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf.fit(
    X_train,
    y_train
)

rf_pred = rf.predict(
    X_test
)

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_precision = precision_score(
    y_test,
    rf_pred,
    average="weighted",
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_pred,
    average="weighted",
    zero_division=0
)

rf_f1 = f1_score(
    y_test,
    rf_pred,
    average="weighted",
    zero_division=0
)

print(
    classification_report(
        y_test,
        rf_pred,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ================================================================
# 33. SVM
# ================================================================

print("\n" + "=" * 75)
print("SUPPORT VECTOR MACHINE")
print("=" * 75)

svm = SVC(
    kernel="rbf",
    C=10,
    gamma="scale"
)

svm.fit(
    X_train_scaled,
    y_train
)

svm_pred = svm.predict(
    X_test_scaled
)

svm_accuracy = accuracy_score(
    y_test,
    svm_pred
)

svm_precision = precision_score(
    y_test,
    svm_pred,
    average="weighted",
    zero_division=0
)

svm_recall = recall_score(
    y_test,
    svm_pred,
    average="weighted",
    zero_division=0
)

svm_f1 = f1_score(
    y_test,
    svm_pred,
    average="weighted",
    zero_division=0
)

print(
    classification_report(
        y_test,
        svm_pred,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ================================================================
# 34. XGBOOST
# ================================================================

print("\n" + "=" * 75)
print("XGBOOST")
print("=" * 75)

number_classes = len(
    np.unique(y)
)

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="mlogloss",
    objective="multi:softprob",
    num_class=number_classes
)

xgb.fit(
    X_train,
    y_train
)

xgb_pred = xgb.predict(
    X_test
)

xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)

xgb_precision = precision_score(
    y_test,
    xgb_pred,
    average="weighted",
    zero_division=0
)

xgb_recall = recall_score(
    y_test,
    xgb_pred,
    average="weighted",
    zero_division=0
)

xgb_f1 = f1_score(
    y_test,
    xgb_pred,
    average="weighted",
    zero_division=0
)

print(
    classification_report(
        y_test,
        xgb_pred,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)


# ================================================================
# 35. MODEL COMPARISON
# ================================================================

results = pd.DataFrame({

    "Model": [
        "Random Forest",
        "SVM",
        "XGBoost"
    ],

    "Accuracy": [
        rf_accuracy,
        svm_accuracy,
        xgb_accuracy
    ],

    "Precision": [
        rf_precision,
        svm_precision,
        xgb_precision
    ],

    "Recall": [
        rf_recall,
        svm_recall,
        xgb_recall
    ],

    "F1 Score": [
        rf_f1,
        svm_f1,
        xgb_f1
    ]
})


print(
    "\n" + "=" * 75
)

print(
    "MODEL PERFORMANCE"
)

print(
    "=" * 75
)

display(
    results
)


# ================================================================
# 36. PERFORMANCE GRAPH
# ================================================================

results_plot = (
    results
    .set_index(
        "Model"
    )
)


results_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "Machine Learning Model Performance"
)

plt.xlabel(
    "Model"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    1.05
)

plt.xticks(
    rotation=0
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ================================================================
# 37. SELECT BEST MODEL
# ================================================================

best_index = (
    results[
        "F1 Score"
    ].idxmax()
)

best_model_name = (
    results.loc[
        best_index,
        "Model"
    ]
)


if best_model_name == "Random Forest":

    best_model = rf

    best_prediction = rf_pred

elif best_model_name == "SVM":

    best_model = svm

    best_prediction = svm_pred

else:

    best_model = xgb

    best_prediction = xgb_pred


print(
    "\nBEST MODEL:",
    best_model_name
)


# ================================================================
# 38. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    best_prediction
)


fig, ax = plt.subplots(
    figsize=(8, 7)
)


disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=label_encoder.classes_
)


disp.plot(
    ax=ax,
    cmap="Blues",
    xticks_rotation=45
)


plt.title(
    "Confusion Matrix - " +
    best_model_name
)

plt.tight_layout()

plt.show()


# ================================================================
# 39. FEATURE IMPORTANCE
# ================================================================

if best_model_name in [
    "Random Forest",
    "XGBoost"
]:

    importance = pd.Series(
        best_model.feature_importances_,
        index=X.columns
    ).sort_values(
        ascending=False
    )


    plt.figure(
        figsize=(10, 6)
    )


    importance.plot(
        kind="bar"
    )


    plt.title(
        "Vibration Feature Importance - " +
        best_model_name
    )


    plt.xlabel(
        "Feature"
    )


    plt.ylabel(
        "Importance"
    )


    plt.xticks(
        rotation=45,
        ha="right"
    )


    plt.grid(
        axis="y",
        alpha=0.3
    )


    plt.tight_layout()

    plt.show()


    print(
        "\nFeature importance:"
    )

    display(
        importance.to_frame(
            "Importance"
        )
    )


# ================================================================
# 40. SAVE MODEL
# ================================================================

model_package = {

    "model":
        best_model,

    "scaler":
        scaler,

    "label_encoder":
        label_encoder,

    "features":
        X.columns.tolist(),

    "sampling_rate":
        SAMPLING_RATE,

    "window_size":
        WINDOW_SIZE
}


joblib.dump(
    model_package,
    "smart_chassis_vibration_model.pkl"
)


# ================================================================
# 41. SAVE FEATURE DATASET
# ================================================================

feature_df.to_csv(
    "smart_chassis_features.csv",
    index=False
)


# ================================================================
# 42. FINAL RESULTS
# ================================================================

print("\n" + "=" * 75)
print("FINAL PROJECT RESULTS")
print("=" * 75)

print(
    "\nDataset: PVS 1 - MPU Left"
)

print(
    "Sampling rate:",
    SAMPLING_RATE,
    "Hz"
)

print(
    "Window duration:",
    WINDOW_SECONDS,
    "seconds"
)

print(
    "Number of features:",
    X.shape[1]
)

print(
    "\nClasses:"
)

print(
    list(
        label_encoder.classes_
    )
)

print(
    "\nRandom Forest Accuracy:",
    round(
        rf_accuracy * 100,
        2
    ),
    "%"
)

print(
    "SVM Accuracy:",
    round(
        svm_accuracy * 100,
        2
    ),
    "%"
)

print(
    "XGBoost Accuracy:",
    round(
        xgb_accuracy * 100,
        2
    ),
    "%"
)

print(
    "\nBest model:",
    best_model_name
)

print(
    "\nSaved files:"
)

print(
    "smart_chassis_vibration_model.pkl"
)

print(
    "smart_chassis_features.csv"
)

print(
    "\n" + "=" * 75
)

print(
    "ML PROJECT COMPLETE"
)

print(
    "=" * 75
)
```
